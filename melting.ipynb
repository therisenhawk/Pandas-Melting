```python
import pandas as pd

df = pd.DataFrame({
    'store': ['A', 'B', 'C'],
    'January': [100, 150, 200],
    'February': [120, 180, 220],
    'March': [140, 160, 250]
})

col_vals = ['January', 'February', 'March']
```

```python
def melt(df, col_vals, key, value):
    keep_vars = [col for col in df.columns if col not in col_vals] #separate the columns you are keeping

    melted = [] #you store up mini dataframes in here

    #loop over the melted values
    for col in col_vals:
        parts = df[keep_vars].copy() #create a small df with kept columns where you add the rest of the info
        parts[key] = col #create a new column and populate it with the melted part
        parts[value] = df[col].values #create a new column and populate it with values from the melted part
        melted.append(parts)

    return pd.concat(melted)

melt(df, col_vals, 'Month', 'Sales')
```
