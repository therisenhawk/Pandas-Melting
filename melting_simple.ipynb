```python
import pandas as pd

df = pd.DataFrame({
    'fruit': ['Apple', 'Banana'],
    'Monday': [10, 20],
    'Tuesday': [15, 25]
})

df
```

```python
#first separate the column/columns you are keeping
piece = df[['fruit']].copy()

#create the column name you want and make every row Monday
piece['day'] = 'Monday'

#Create a new column called sales and fills every row with 'Monday' values.
piece['sales'] = df['Monday'].values

#now repeat for other days

piece2 = df[['fruit']].copy()

piece2['day'] = 'Tuesday'

piece2['sales'] = df['Tuesday'].values

final_data = pd.concat([piece, piece2], ignore_index=True)
final_data
```
